## SQLite3

```python
import sqlite3
```

```python
connection = sqlite3.connect("example.db")
cursor = connection.cursor()
print(connection, cursor)
```

```python
cursor.execute('''
Create table if not Exists employees(id Integer Primary Key, name Text Not Null, age Integer, department Text)
''')
connection.commit()
```

```python
# Insert the data
cursor.execute('''
    Insert into employees(name, age, department) values('Sandeep', 32, 'Nic Officer')
''')

cursor.execute('''
    Insert into employees(name, age, department) values('Shivam', 22, 'Cyber Sec Expert')
''')

cursor.execute('''
    Insert into employees(name, age, department) values('Vishal', 12, 'Gaming Expert')
''')

connection.commit()
```

```python
# Querying the data
cursor.execute('''
    Select * from employees
''')
rows = cursor.fetchall()
rows
```

```python
cursor.execute('''
    UPDAte employees set age=34 where name="Sandeep"
''')
connection.commit()
```
