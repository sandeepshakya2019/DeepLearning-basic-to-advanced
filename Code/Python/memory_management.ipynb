# GC Module

```python
import gc
```

```python
class Obj:
    def __init__(self, name):
        self.name = name
        print("Object is created", self.name)
    
    def __del__(self):
        print(f"Obbject is deleted {self.name}")

# circular refrence

obj1 = Obj("sandeep")
obj2 = Obj("shakya")

obj1.ref = obj2
obj2.ref = obj1

# del obj1
# del obj2

gc.collect()
```

# Genbrators for memory efficiencey

```python
def genrate(n):
    for i in range(n):
        yield i

for num in genrate(100):
    print(num)
```
