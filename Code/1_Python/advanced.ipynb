```python
import logging

# Global configuration
logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s - %(levelname)s - %(name)s - %(message)s',
    datefmt='%H:%M:%S - %d-%m-%Y',
    force=True,
    handlers=[
        logging.FileHandler("app.log"),
        logging.StreamHandler()
    ]
)

logger = logging.getLogger("AirthmeticApp")
logger.setLevel(logging.DEBUG)
```

```python
def add(a, b):
    try:
        result = a + b
        logger.debug(f"Adding {a} + {b} = {result}")
        return result
    except TypeError as e:
        logger.error(f"Type error while adding: {e}")
        return None
    except Exception as e:
        logger.exception(f"Unexpected error in add(): {e}")
        return None
    finally:
        logger.info("add() function execution completed.")
```

```python
def divide(a, b):
    try:
        result = a / b
        logger.debug(f"Dividing {a} / {b} = {result}")
        return result
    except ZeroDivisionError:
        logger.error("Division by zero error")
        return None
    except TypeError as e:
        logger.error(f"Type error: {e}")
        return None
    except Exception as e:
        logger.exception(f"Unexpected error: {e}")
        return None
    finally:
        logger.info("divide() function execution completed.")
```

```python
print("Result 1:", add(10, 5))         # ✅ 15
print("Result 2:", add(10, "a"))       # ❌ TypeError
print("Result 3:", divide(20, 4))      # ✅ 5.0
print("Result 4:", divide(5, 0))       # ❌ ZeroDivisionError
print("Result 5:", divide("x", 2))     # ❌ TypeError
```

```python
import threading
import time
```

```python
import time

res1 = 0
res2 = 0

def print_numbers():
    for i in range(10000000):
        global res1
        res = i ** 2 + i ** 3 + i ** 2
        res1 += res

def print_letter():
    for i in range(10000000):
        global res2
        res = i ** 2 + i ** 3 + i ** 2
        res2 += res
```

```python
start = time.time()
print_numbers()
print_letter()
end = time.time()

elapsed_ms = (end - start) * 1000
elapsed_sec = elapsed_ms / 1000
print(f"Time taken (seconds): {elapsed_sec:.4f} and res1 : {res1} , res2 : {res2}")
```

```python


thread1 = threading.Thread(target=print_numbers)
thread2 = threading.Thread(target=print_letter)

start = time.time()
thread1.start()
thread2.start()

# thread1.join()
# thread2.join()

end = time.time()


elapsed_ms = (end - start) * 1000
elapsed_sec = elapsed_ms / 1000
print(f"Time taken (seconds): {elapsed_sec:.4f} and res1 : {res1} , res2 : {res2}")
```

```python
from multiprocessing import Process
import time

def task(name):
    print(f"Task {name} started", flush=True)
    time.sleep(1)
    print(f"Task {name} finished", flush=True)

if __name__ == "__main__":
    p1 = Process(target=task, args=("A",))
    p2 = Process(target=task, args=("B",))

    p1.start()
    p2.start()

    p1.join()
    p2.join()

    print("Both processes completed", flush=True)
```

```python
from concurrent.futures import ThreadPoolExecutor
import time

def task(name):
    print(f"Task {name} started")
    time.sleep(1)
    print(f"Task {name} finished")
    return f"Result from {name}"

if __name__ == "__main__":
    start = time.time()

    with ThreadPoolExecutor(max_workers=4) as executor:
        names = ["A", "B", "C", "D"]
        results = executor.map(task, names)  # Run tasks concurrently

    end = time.time()
    print("Results:", list(results))
    print(f"Total time: {end - start:.2f} seconds")
```

```python
from concurrent.futures import ThreadPoolExecutor
import requests
from bs4 import BeautifulSoup
import re
import time

# Start URLs for crawling
start_urls = [
    "https://www.iittp.ac.in/",
    "https://www.iittp.ac.in/contact-us",
    "https://www.iittp.ac.in/faculty",
]

email_pattern = re.compile(r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}')
headers = {'User-Agent': 'Mozilla/5.0 (compatible; EmailScraper/1.0)'}

def scrape_emails(url):
    print(f"Scraping: {url}")
    try:
        response = requests.get(url, headers=headers, timeout=10)
        response.raise_for_status()
    except Exception as e:
        print(f"⚠️ Failed: {url} → {e}")
        return url, set()

    soup = BeautifulSoup(response.text, "html.parser")
    text = soup.get_text()
    emails = set(re.findall(email_pattern, text))
    print(f"✅ {len(emails)} emails found at {url}")
    return url, emails

def crawl_and_extract(start_urls):
    start = time.time()
    with ThreadPoolExecutor(max_workers=5) as executor:
        results = executor.map(scrape_emails, start_urls)

    all_emails = {}
    for url, emails in results:
        all_emails[url] = emails

    elapsed = time.time() - start
    print(f"\n⏱ Total Time: {elapsed:.2f} seconds\n")
    return all_emails

if __name__ == "__main__":
    found = crawl_and_extract(start_urls)
    for url, emails in found.items():
        print(f"{url} → {emails if emails else 'No emails found'}")
```
