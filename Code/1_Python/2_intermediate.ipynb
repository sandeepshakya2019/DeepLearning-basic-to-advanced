### Inheritance

```python
class Car:
    def __init__(self, windows, doors, enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype

    def drive(self):
        print(f"Person is driving the {self.enginetype} car")

class Tesla(Car):
    def __init__(self, windows, doors, enginetype, isSelfDrive):
        super().__init__(windows, doors, enginetype)
        self.isSelfDrive = isSelfDrive

    def selfDrive(self):
        if self.isSelfDrive:
            print(f"Tesla is drive self car with engine {self.enginetype}")
        else:
            print(f"Tesla is no using self drive car with engine {self.enginetype}")

t1 = Tesla(4, 4, "fast", True)
t2 = Tesla(4, 4, "slow", False)

t1.selfDrive(), t2.selfDrive()



```

### Mulitple inheritance

```python
class Father:
    def __init__(self):
        print("Father's __init__")

class Mother:
    def __init__(self):
        print("Mother's __init__")

class Child(Father, Mother):
    def __init__(self):
        super().__init__()  # Only calls Father's __init__ due to MRO
        print("Child's __init__")

c = Child()
```

### Polymorphism

```python
class Animal:
    def speak(self):
        return "Sound of animal"
    
class Dog(Animal):
    def speak(self):
        return "Woof !!"
    
class Cat(Animal):
    def speak(self):
        return "Meow !!"
    
d = Dog()
c = Cat()
print(d.speak(), c.speak())
```

```python
class Shape:
    def area():
        return
    def parameter():
        return
    
class Rectangle(Shape):
    def __init__(self, width, height):
        super().__init__()
        self.width = width
        self.height = height

    def area(self):
        print(self.height * self.width)
    
    def parameter(self):
        print(2 * (self.height + self.width))

class Square(Shape):
    def __init__(self, a):
        super().__init__()
        self.width = a

    def area(self):
        print(self.width * self.width)
    
    def parameter(self):
        print(4 * self.width)

s = Square(5)
s.area()
s.parameter()

s = Rectangle(5, 6)
s.area()
s.parameter()
```

### Polymorphism abstract base class

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        # return super().start_engine()
        print("Card Starts ")
    
class Bike(Vehicle):
    def start_engine(self):
        # return super().start_engine()
        print("Bike Starts ")

Car().start_engine()
Bike().start_engine()
```

### Magic Methods

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name} and {self.age}"
    
    def __add__(self, otherclass):
        return f"{self.name} & {otherclass.name}"

p1 = Person("sandeep", 56)
p2 = Person("Spme", 56)

p1.__str__()
p1.__add__(p2)
p1.__sizeof__()
```
