## Reading and Displaying Images with OpenCV
This guide explains the fundamental process of loading an image from a file and displaying it on the screen using OpenCV's core functions. It's one of the first steps in any computer vision project!

Core Functions You'll Need
You only need a few essential functions to get started:

Function

Description

📂 cv2.imread()

Reads an image from a specified file path. It returns the image as a NumPy array. If the image cannot be read (e.g., wrong path), it returns None.

🖥️ cv2.imshow()

Displays an image in a window. The window automatically fits to the image size. The first argument is the window's title (a string), and the second is the image you want to show.

⌨️ cv2.waitKey()

Waits for a specified number of milliseconds for a keyboard event. If you pass 0 as the argument, it will wait indefinitely until any key is pressed. This is essential for keeping the image window open.

❌ cv2.destroyAllWindows()

Closes all the windows created by OpenCV. It's good practice to call this at the end to clean up resources.

⚠️ The Golden Rule: Always Check if the Image Loaded
The most common error when working with images is providing an incorrect file path. cv2.imread() does not throw an error in this case; it simply returns None. If you then pass this None object to cv2.imshow(), your program will crash.

Therefore, you must always check if the image object is None right after calling cv2.imread().

## Reading Videos

To read a video or webcam feed with OpenCV:

1.  **Create a capture object:**
    * For a file: `cap = cv2.VideoCapture('path/to/video.mp4')`
    * For a webcam: `cap = cv2.VideoCapture(0)`

2.  **Loop and read frames:**
    * Use a `while` loop to read the video frame by frame with `ret, frame = cap.read()`.
    * `ret` will be `False` when the video ends.

3.  **Display the frame:**
    * Use `cv2.imshow('Window Name', frame)` to show the image.

4.  **Release resources:**
    * When you're done, call `cap.release()` and `cv2.destroyAllWindows()` to free the camera/file and close the display window.

```python
v1 = f"{video_path}/dog.mp4"
capture = cv.VideoCapture(v1)
# capture = cv.VideoCapture(1)
```

```python
# isTrue, frame = capture.read()
# # frame.shape
# cv.imshow("FirstFrame", frame)
# cv.waitKey(0)
```

```python
# Assuming 'capture = cv.VideoCapture(...)' is defined before this loop

while True:
    # Read a frame
    isTrue, frame = capture.read()
    # print(isTrue)
    # IMPORTANT: Check if a frame was successfully read
    if isTrue:
        # If yes, display it
        cv.imshow('video', frame)

        # Check if the 'q' key was pressed to exit
        if cv.waitKey(20) & 0xFF == ord('q'):
            break
    else:
        # If isTrue is False, it means the video has ended.
        # Break the loop.
        break

# Release the capture and destroy windows
capture.release()
cv.destroyAllWindows()
```

## Capture live camera

```python
import cv2 as cv

# To capture from a webcam, pass the camera's index to VideoCapture.
# 0 is usually the default built-in webcam.
# If you have multiple cameras, you can try 1, 2, etc.
capture = cv.VideoCapture(0)

# Always check if the camera was opened successfully.
if not capture.isOpened():
    print("Error: Cannot open camera. Make sure it is connected and not in use by another application.")
else:
    # Loop indefinitely to capture frames in real-time.
    while True:
        # Read one frame from the camera.
        # ret is a boolean: True if the frame was read successfully.
        # frame is the actual image data.
        ret, frame = capture.read()

        # If a frame was successfully captured...
        if ret:
            # Display the resulting frame in a window titled 'Live Webcam Feed'.
            cv.imshow('Live Webcam Feed', frame)

            # Wait for 1 millisecond and check if the 'q' key is pressed.
            # If 'q' is pressed, break out of the loop to stop the feed.
            # Using waitKey(1) is essential for a smooth, real-time video feed.
            if cv.waitKey(1) & 0xFF == ord('q'):
                break
        else:
            # If ret is False, it means there was an error capturing the frame.
            print("Error: Can't receive frame (stream end?). Exiting ...")
            break

# When the loop is finished, release the camera resource.
# This is a crucial step to free up the camera for other applications.
capture.release()

# Destroy all the windows created by OpenCV.
cv.destroyAllWindows()
```

## resizing and reascaling

```python
import cv2 as cv
faces_path = "Resources/Faces"
video_path = "Resources/Videos"
photos_path = "Resources/Photos"
```

```python
import cv2 as cv

# --- Functions for Resizing and Rescaling ---

def rescaleFrame(frame, scale=0.75):
    """
    Resizes a frame to a specific scale.
    Works for images, videos, and live video.
    """
    # Get the original dimensions
    width = int(frame.shape[1] * scale)
    height = int(frame.shape[0] * scale)
    dimensions = (width, height)

    # Resize the frame
    return cv.resize(frame, dimensions, interpolation=cv.INTER_AREA)

def changeRes(capture, width, height):
    """
    Changes the resolution of the capture feed.
    Only works for live video from a webcam.
    """
    capture.set(3, width)  # 3 corresponds to CAP_PROP_FRAME_WIDTH
    capture.set(4, height) # 4 corresponds to CAP_PROP_FRAME_HEIGHT
```

```python
capture_video = cv.VideoCapture(f"{video_path}/dog.mp4")
while True:
    isTrue, frame = capture_video.read()
    if isTrue:
        frame_resized = rescaleFrame(frame)
        cv.imshow("Video", frame)
        cv.imshow("ResizeVideo", frame_resized)

        if cv.waitKey(1) & 0xFF == ord('q'):
            break
    else:
        break
capture_video.release()
cv.destroyAllWindows()
```

```python
# draw a rectabg;le
blank = np.zeros(img.shape, dtype='uint8')

cv.rectangle(blank, pt1 = (0,0), pt2=(250,400), color=(0,255,0), thickness=2)
cv.rectangle(blank, pt1 = (36,56), pt2=(1,1), color=(0,255,0), thickness=cv.FILLED)
cv.rectangle(blank, pt1 = (80,80), pt2=(90,90), color=(0,255,0), thickness=-1)


cv.imshow('img', blank)
cv.waitKey(0)
```

```python
# draw a rectabgle and circle and circle and text
blank = np.zeros((500,500, 3), dtype='uint8')

cv.rectangle(blank, pt1=(0,0), pt2=(blank.shape[0]//2, blank.shape[1]//2), color=(0,255,0), thickness=2)

cv.circle(blank, center = (blank.shape[0]//4, blank.shape[1]//4), radius=blank.shape[1]//4, color=(0,255,0), thickness=-1)

cv.line(blank, pt1=(0,0), pt2=(blank.shape[0]//2, blank.shape[1]//2), color=(0,0,255), thickness=2)
cv.line(blank, pt1=(0,blank.shape[0]//2), pt2=(blank.shape[0]//2, 0), color=(0,0,255), thickness=2)

cv.putText(blank, text='Line text', org=(255,255), fontFace=cv.FONT_HERSHEY_COMPLEX, fontScale=1, color=(0,255,2), thickness=2)

cv.imshow('img', blank)
cv.waitKey(0)
```
