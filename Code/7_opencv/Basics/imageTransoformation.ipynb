```python
def translate(img, x, y):
    """
    Translates (shifts) an image horizontally and vertically.

    Args:
        img: The input image (as a NumPy array).
        x: The number of pixels to shift along the X-axis (horizontal).
           A positive value shifts the image right, a negative value shifts it left.
        y: The number of pixels to shift along the Y-axis (vertical).
           A positive value shifts the image down, a negative value shifts it up.

    Returns:
        The newly translated image.
    """
    # Define the translation matrix. This matrix is the "recipe" for the shift.
    # [1, 0, x] means the new x-coordinate is 1*old_x + 0*old_y + x.
    # [0, 1, y] means the new y-coordinate is 0*old_x + 1*old_y + y.
    transMat = np.float32([[1, 0, x], [0, 1, y]])

    # Define the dimensions (width, height) of the output image.
    # This should match the original image's dimensions.
    # img.shape[1] is width, img.shape[0] is height.
    dimensions = (img.shape[1], img.shape[0])

    # Use OpenCV's warpAffine function to apply the translation.
    # It takes the original image, the transformation matrix, and the output size.
    return cv.warpAffine(img, transMat, dimensions)
```

```python
img = cv.imread("Resources/Photos/cat.jpg")
translate_img = translate(img, -100, 100)
cv.imshow("Original Img", img)
cv.imshow("translate_img", translate_img)

cv.waitKey(0)
cv.destroyAllWindows()
```

```python
def scaledImg(img, s_x = 2.0, s_y = 0.5):
    scaleMat = np.float32([[s_x, 0, 0], [0, s_y, 0]])
    dimensions = (img.shape[1], img.shape[0])
    return cv.warpAffine(img, scaleMat, dimensions)
```

```python
img = cv.imread("Resources/Photos/cat.jpg")

translate_img = translate(img, -200, 100)
scaled_img = scaledImg(translate_img, 2, 0.5)
cv.imshow("Original Img", translate_img)
cv.imshow("translate_img", scaled_img)

cv.waitKey(0)
cv.destroyAllWindows()
```

```python
def rotateImage(img, rotPoint = None, angle = 90, scale = 1.0):
    # Get image dimensions
    (height, width) = img.shape[:2]

    # Define the center of rotation and the angle (in degrees)
    if rotPoint == None:
        center = (width // 2, height // 2)
    else:
        center = rotPoint

    angle = angle  # Rotate by 45 degrees
    scale = scale # Keep the original size

    # Get the rotation matrix
    rotMat = cv.getRotationMatrix2D(center, angle, scale)

    # Now, you can use this matrix with warpAffine
    # rotated_img = cv.warpAffine(img, rotMat, (width, height))
    dimensions = (img.shape[1], img.shape[0])
    
    return cv.warpAffine(img, rotMat, dimensions)
    
```

```python
rotate_img = rotateImage(img, rotPoint=(90,90))
cv.imshow("img", img)
cv.imshow("rotate", rotate_img)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
def resize_flexible(img, width=None, height=None, scale_percent=None):
    """
    Resizes an image by width, height, or percentage.
    Only one argument should be provided.
    """
    h, w = img.shape[:2]

    if scale_percent:
        new_w = int(w * scale_percent / 100)
        new_h = int(h * scale_percent / 100)
    elif width:
        new_w = width
        new_h = int(h * (width / w)) # Maintain aspect ratio
    elif height:
        new_h = height
        new_w = int(w * (height / h)) # Maintain aspect ratio
    else:
        return img # No change

    # Choose interpolation method based on scaling up or down
    inter = cv.INTER_AREA if new_w < w else cv.INTER_LINEAR
    
    return cv.resize(img, (new_w, new_h), interpolation=inter)
```

```python
def resizeImage(img, pixels = None):
    # If no width is specified, return the original image
    if pixels is None:
        return img

    # Get the original dimensions
    (h, w) = img.shape[:2]
    
    # Calculate the aspect ratio and the new height
    aspect_ratio = h / w
    new_height = int(pixels * aspect_ratio)
    # New dimensions tuple (width, height)
    dimensions = (pixels, new_height)
    
    # Resize the image using cv.resize
    # cv.INTER_AREA is generally best for shrinking images.
    return cv.resize(img, dimensions, interpolation=cv.INTER_CUBIC)
```

```python
rotate_img = resizeImage(img, pixels=900)
cv.imshow("img", img)
cv.imshow("rotate", rotate_img)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
import cv2 as cv

img = cv.imread("Resources/Photos/cat.jpg")
```

```python
import cv2 as cv

# Assuming 'img' has been loaded, for example:
# img = cv.imread('your_image.jpg')

# Convert the original color image (BGR) to grayscale.
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

# Apply the Canny edge detection algorithm to find edges in the grayscale image.
# 125 and 175 are the lower and upper threshold values.
canny = cv.Canny(gray, 125, 175)

# Display the original color image in a window titled 'Cats'.
cv.imshow('Cats', img)

# Display the grayscale image in a window titled 'Gray'.
cv.imshow('Gray', gray)

# Display the image with detected edges in a window titled 'CAnny'.
cv.imshow('CAnny', canny)

# Wait indefinitely for a key to be pressed before proceeding.
cv.waitKey(0)

# Close all the open OpenCV windows.
cv.destroyAllWindows()
```
